```python
import numpy as np
from matplotlib.ticker import EngFormatter
```

# CCM

```python
# Parameters
Vs = 170
D = 0.5
N1 = 5
N2 = 1
R = 25
fs = 77e3
# Output voltage calculation for CCM
Vo = (Vs*D*N2)/(N1*(1-D))
print(f"Output Voltage (V_o): {EngFormatter(unit='V').format_data(Vo)}")

# Continuous Conduction Mode (CCM) boundary inductance calculation
Lccm = ((R*(1-D)**2)/(2*fs))*((N1/N2)**2)
print(f"CCM Boundary Inductance (L_ccm): {EngFormatter(unit='H').format_data(Lccm)}")


```

# DCM

```python
Ldcmm = 62.5e-6

Vo = (N2/N1)*D*Vs*np.sqrt(R/(2*Ldcmm*fs))
print(f"Output Voltage at DCM boundary (V_o): {EngFormatter(unit='V').format_data(Vo)}")

```
